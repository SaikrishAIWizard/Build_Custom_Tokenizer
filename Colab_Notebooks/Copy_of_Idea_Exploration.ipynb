```python
from transformers import PreTrainedTokenizerFast

tok = PreTrainedTokenizerFast.from_pretrained("ashutran63/medical-bpe-16k")

text = "Acetylcholinesterase inhibitors are used in Alzheimer's disease treatment."
tokens = tok.tokenize(text)
print(tokens)
```

```python
from transformers import PreTrainedTokenizerFast

tok = PreTrainedTokenizerFast.from_pretrained("paulvision/medical-bpe-pubmed")

text = "Acetylcholinesterase inhibitors are used in Alzheimer's disease treatment."
tokens = tok.tokenize(text)
print(tokens)
```

```python
ids = tok.encode(text)
print(tok.decode(ids))
```

```python
encoding = tok(text)

print(encoding["input_ids"])
print(tok.convert_ids_to_tokens(encoding["input_ids"]))
```

```python
print("Vocabulary size:", tok.vocab_size)
print("Special tokens:", tok.all_special_tokens)
print("Model max length:", tok.model_max_length)
```
